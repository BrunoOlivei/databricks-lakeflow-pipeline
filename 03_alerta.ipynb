# 03 · Alerta (task do job)

Notebook genérico de alerta, usado em dois lugares do job:

| Task | Como dispara | Parâmetro `mensagem` |
|---|---|---|
| `alerta_sem_pedidos` | Ramo **false** do If/else (`pedidos_ultimo_dia > 0` falso) | `Gold sem pedidos no último dia ({{tasks.checagem_gold.values.ultima_data}})` |
| `alerta_falha` | **Run if = At least one failed**, dependendo da pipeline e da checagem | `Falha no job {{job.name}}, execução {{job.run_id}}` |

Num projeto real, aqui entraria um webhook (Teams/Slack). Para estudo, ele só registra e, se `falhar = sim`, termina com erro para você ver a task vermelha no DAG.

```python
dbutils.widgets.text("mensagem", "Alerta sem mensagem")
dbutils.widgets.dropdown("falhar", "nao", ["nao", "sim"])
 
mensagem = dbutils.widgets.get("mensagem")
print(f"[ALERTA] {mensagem}")
 
if dbutils.widgets.get("falhar") == "sim":
    raise RuntimeError(mensagem)
```
